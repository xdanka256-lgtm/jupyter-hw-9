```python
import numpy as np 
import pandas as pd 
from IPython.display import display 
from sympy import symbols, Eq, solve
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error, r2_score
```

```python
print("\033[35mTask number 1\033[0m")

def first_linear_function(x):
    return 12 - x

def second_linear_function(x):
    return x - 4

x_values = np.linspace(-15, 35, 100)
y_values_1 = first_linear_function(x_values)
y_values_2 = second_linear_function(x_values)

plt.plot(
    x_values,
    y_values_1,
    label = "x + y = 12"
)

plt.plot(
    x_values,
    y_values_2,
    label = "x - y = 4"
)

plt.axhline(
    y=0,
    color="black",
   linewidth=0.8,
    linestyle="--"
)
plt.axvline(
    x=0,
    color="black",
    linewidth=0.8,
    linestyle="--"
)

plt.title("Linear equations")
plt.xlabel("x")
plt.ylabel("y")

plt.xlim(-10,20)
plt.ylim(-20,20)

plt.grid(
    color="gray",
    linestyle="--",
    linewidth=0.5
)

plt.legend()
plt.show()
```

```python
print("\033[35mTask number 2\033[0m")

x,y = symbols("x,y")

equation_1 = Eq(x+y,12)
equation_2 = Eq(x-y,4)

solution = solve(
    (equation_1,equation_2),
    (x,y)
)

print(solution)
intersection_x=float(solution[x])
intersection_y=float(solution[y])

x_values=np.linspace(-10,20,100)

y_values_1 = 12 - x_values
y_values_2 = x_values - 4 

plt.plot(
    x_values,
    y_values_1,
    label = "x + y = 12"
)

plt.plot(
    x_values,
    y_values_2,
    label = "x - y = 4"
)

plt.scatter(
    intersection_x,
    intersection_y,
    color="red",
    s=80,
    label="Intersection"
)

plt.annotate(
    f"({intersection_x:.0f},{intersection_y:.0f})",
    xy=(intersection_x,intersection_y),
    xytext=(intersection_x+1,intersection_y+2),
    arrowprops={"arrowstyle":"->"}
)

plt.axhline(
    y=0,
    color="black",
    linewidth=0.8,
    linestyle="--"
)
plt.axvline(
    x=0,
    color="black",
    linewidth=0.8,
    linestyle="--"
)
plt.title("Interception of 2 linear equations")
plt.xlabel("x")
plt.ylabel("y")

plt.xlim(-10,20)
plt.ylim(-20,20)

plt.grid(
    color="gray",
    linestyle="--",
    linewidth=0.5
)

plt.legend()
plt.show()
```
